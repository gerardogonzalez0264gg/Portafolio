# Convertir columnas de fecha serial de Excel a fecha normal

Este notebook:
1. Carga el archivo `query_chile.xlsx`
2. Convierte las columnas `time` y `updated` desde formato serial de Excel a fecha/hora
3. Exporta un nuevo archivo Excel corregido


```
import pandas as pd
```

```
# 1) Cargar archivo
archivo_entrada = "query_chile.xlsx"
df = pd.read_excel(archivo_entrada)

print("Columnas del archivo:")
print(df.columns.tolist())
print("\nPrimeras filas:")
display(df.head())
```

```
# 2) Convertir columnas seriales de Excel a fecha/hora normal
# Excel cuenta los días desde 1899-12-30
columnas_fecha = ["time", "updated"]

for col in columnas_fecha:
    if col in df.columns:
        df[col] = pd.to_datetime(df[col], unit="D", origin="1899-12-30", errors="coerce")

print("Tipos de datos después de la conversión:")
print(df.dtypes)

print("\nMuestra de las columnas convertidas:")
cols_mostrar = [c for c in ["time", "updated"] if c in df.columns]
display(df[cols_mostrar].head())
```

```
# 3) Exportar archivo corregido
archivo_salida = "query_chile_corregido.xlsx"
df.to_excel(archivo_salida, index=False)

print(f"Archivo exportado correctamente: {archivo_salida}")
```
